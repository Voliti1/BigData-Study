```python
from SQLite import *

SL = SQLite()

sql = 'select "호선명", "승객유형", "07-08시간대", "08-09시간대" from subway'
result = SL.run_sql(sql)
print(result.head(10), '\n')

df_pivot = result.pivot_table(index = "호선명", columns = "승객유형", values = "07-08시간대", aggfunc = 'sum' )
print(df_pivot)

df_pivot.to_csv('subway_summary.csv', encoding = "EUC-KR")
SL.close_db()
```
