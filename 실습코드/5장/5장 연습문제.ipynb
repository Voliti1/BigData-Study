```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

#1
df = sns.load_dataset('tips')
print(df, '\n')
day = df['day']

#2
print(day, '\n')
print(day.dtype, '\n')

#3
fd = day.value_counts(sort = False)
print(fd / len(day))

#4
print(fd, '\n')

#5
fd.plot.bar(xlabel = 'Day', ylabel = 'Frequency', rot = 0, title = 'Frequency of days')
plt.show()

#6
fd.plot.barh(xlabel = 'Frequency', ylabel = 'Day', rot = 0, title = 'Frequency of days')
plt.show()

#7
exp = [0, 0.1, 0, 0.05]
wedge = {'edgecolor' : 'black', 'linewidth' : 5, 'width' : 0.1}
fd.plot.pie(xlabel = 'Frequency', title = 'Frequency of days', autopct = '%.1f%%', explode = exp, counterclock = False, wedgeprops = wedge)
plt.show()

#8
smoke = df['smoker'].value_counts()
color = ['lightblue', 'lightgreen']
smoke.plot.bar(xlabel = 'Smoker', ylabel = 'Num', rot = 0, title = "Number of smoker", ylim = [0, 200], color = color)
plt.show()

#9
time = df.time.value_counts()
time.plot.pie(xlabel = "Frequency", autopct = '%.1f%%', title = "Lunch or Dinner")
plt.show()
```
