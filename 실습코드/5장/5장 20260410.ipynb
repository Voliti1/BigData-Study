```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/cars.csv");
dist = df['dist']
print(dist.describe())

dist.plot.hist()
plt.show()

dist.plot.hist(bins = 6)
plt.show()

print(dist.value_counts(bins = 6, sort = False))

dist.plot.hist(bins = 6, 
               title = 'Braking Distance', 
               xlabel = 'Distance', 
               ylabel = 'Frequency', 
               color = 'g',
               edgecolor = 'black')
plt.show()

dist.plot.box(title='Breaking Distance')
print(dist.describe())
plt.show()

df.boxplot(column = "dist", by = 'speed', grid = False)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rc('font', family='Malgun Gothic')
plt.rc('axes', unicode_minus=False)

# 0~2π 범위 생성

x = np.linspace(0, 2 * np.pi, 100)
plt.figure(figsize=(8, 5))

# 동일 x 범위에서 두 함수를 그리면 직접 비교 가능

plt.plot(x, np.sin(x), label="sin(x)", linewidth=2)
plt.plot(x, np.cos(x), label="cos(x)", linewidth=2)

plt.title("sin(x)와 cos(x) 비교 그래프")
plt.xlabel("x (라디안)")
plt.ylabel("함수 값")
plt.legend() # 그래프 구분을 위한 범례
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/iris.csv")
fig, axes = plt.subplots(nrows = 2, ncols = 2)

df['Petal_Length'].plot.hist(ax = axes[0, 0])
df['Petal_Length'].plot.box(ax = axes[0, 1])

fd = df['Species'].value_counts()
fd.plot.pie(ax = axes[1, 0])
fd.plot.barh(ax = axes[1, 1])

fig.suptitle('Multiple Graph Example', fontsize = 14)
plt.show()
```
