```python
import numpy as np
import pandas as pd
import seaborn as sns
from scipy import stats
from sklearn.impute import KNNImputer
from sklearn.preprocessing import MinMaxScaler

np.set_printoptions(precision=6, suppress=True)
pd.options.display.float_format = '{:,.6f}'.format

df = sns.load_dataset("planets")

#1
print("1번")
print(df.isna().sum(), '\n')

#2
print("2번")
print(df.isna().sum().sum(), '\n')

#3
print("3번")

df1 = df.dropna()
df1 = df1.reset_index(drop = True)
df_clean = df1.loc[:, ~df1.columns.isin(['mass', 'distance'])]
print(df_clean, '\n')

#4
print("4번")

df2 = df.drop(columns=['method'])
original_columns = df2.columns
print(df2['orbital_period'].mean())

scaler = MinMaxScaler()
df_scaled = scaler.fit_transform(df2)

imputer = KNNImputer(n_neighbors = 5)
df_scaled = imputer.fit_transform(df_scaled)

df_filled = scaler.inverse_transform(df_scaled)

df2 = pd.DataFrame(df_filled, columns = original_columns)

print(df2['orbital_period'].mean(), '\n')

#5
print("5번")

z = np.abs(stats.zscore(df2['orbital_period']))
outliers = df2['orbital_period'][z > 2]
print(outliers, '\n')

#6
print("6번")

Q1 = df['distance'].quantile(0.25)
Q3 = df['distance'].quantile(0.75)
IQR = Q3 - Q1

outliers = df['distance'][(df['distance'] < Q1 - IQR * 1.5) | (df['distance'] > Q3 + IQR * 1.5)]
print(outliers, '\n')

#7
print("7번")
dist_clean = df['distance']
dist_clean = dist_clean.loc[~dist_clean.isin(outliers)]
print(dist_clean, '\n')

#8
print("8번")
df.sort_values("mass", ascending = False)
print(df, '\n')

#9
print("9번")
rank = df['mass'].rank(ascending = False)
print(rank, '\n')

#10
print("10번")
df20 = df.sample(n = 20, random_state = 123)
print(df20)
```

```python
import numpy as np
import pandas as pd
import seaborn as sns

pd.options.display.float_format = '{:,.0f}'.format

df = sns.load_dataset('penguins')
df = df.dropna()
print(df, '\n')

#11
print("11번")

print(df.groupby('species')['bill_length_mm'].mean(), '\n')
print(df.groupby('species')['bill_depth_mm'].mean(), '\n')

#12
print("12번")

print(df.groupby('sex')['body_mass_g'].max(), '\n')
print(df.groupby('sex')['flipper_length_mm'].max(), '\n')

#13
print("13번")

print(df.groupby('sex')['bill_length_mm'].mean(), '\n')
print(df.groupby('sex')['bill_depth_mm'].mean(), '\n')

#14
print("14번")

print(df.groupby(['species', 'sex']).value_counts(), '\n')

#15
print("15번")

print(df.groupby(['species', 'island'])['flipper_length_mm'].mean(), '\n')

#16
print("16번")

print(df.groupby(['species', 'sex'])['body_mass_g'].mean(), '\n')
```
