```python
import pandas as pd
import matplotlib.pyplot as plt

favorite = pd.Series(['WINTER', 'SUMMER', 'SPRING', 'SUMMER', 'SUMMER', 'FALL', 'FALL', 'SUMMER', 'SPRING', 'SPRING'])

print(favorite, '\n')
print(favorite.value_counts(), '\n')
print(favorite.value_counts() / favorite.size, '\n')

fd = favorite.value_counts()
print(type(fd), '\n')
print(fd['SUMMER'], '\n')
print(fd.iloc[2], '\n')


fig, axes = plt.subplots(2, 2, figsize=(15, 10))

fd.plot.bar(xlabel = 'Season', ylabel = 'Frequency', rot = 0, title = 'Favorite Season', ax = axes[0, 0])

fd.plot.barh(xlabel = 'Frequency', ylabel = 'Season', rot = 0, title = 'Favorite Season', ax = axes[0, 1])

fd.plot.pie(ylabel = '', autopct = '%1.0f%%', title = 'Favorite Season', ax = axes[1, 0])

plt.subplots_adjust(left = 0.2)
plt.suptitle('Favorite Season Graph')
plt.show()

```
