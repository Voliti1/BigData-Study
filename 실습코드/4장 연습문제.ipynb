```python
import seaborn as sns
import pandas as pd

df = sns.load_dataset('titanic')

#9
print(df.loc[(df.alive == "yes") & (df.sex == "male"), 'age'].mean(), '\n')

#10
print(df.loc[df.alive == 'yes', 'alive'].count() / len(df.alive), '\n')

#11
print(df.loc[df.age == min(df.age), ['sex', 'age']], '\n')

#12
print(df.loc[(df.pclass == 1) & (df.alive == 'yes'), 'alive'].count(), '\n')

#13
print(df.loc[df.pclass == 1, 'pclass'].count() - df.loc[df.pclass == 2, 'pclass'].count(), '\n')

#14 *
a = df.loc[(df.pclass == 3) & (df.alive == 'yes'), ['sex', 'age']]
print(a.loc[a.age == a.age.max()], '\n')

# print(df.loc[(df.pclass == 3) & (df.alive == 'yes') & (df.age == df.loc[(df.pclass == 3) & (df.alive == 'yes'), 'age'].max()), ['sex', 'age']])

#15
df2 = df[['pclass', 'sex', 'age', 'alive']].copy()
print(df2, '\n')

#16
df2['age_class'] = None
df2.loc[df2.age < 40, 'age_class'] = 'Low'
df2.loc[df2.age >= 40, 'age_class'] = 'High'
df2['age_class'] = df2['age'].apply(lambda x: 'Low' if x < 40 else ('High' if x >= 40 else None))

print(df2, '\n')

#17
print(df2.age + 10, '\n')

#18
print((df2.age + 10).mean(), '\n')

#19 *
print(df2.loc[:, df2.columns != 'pclass'], '\n')

#20
df2.to_csv("titanic.csv")
```
