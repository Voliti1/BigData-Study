```python
import pandas as pd

age = pd.Series([25, 34, 19, 45, 60])
print(age, '\n')
print(type(age), '\n')

data = ['Spring', 'Summer', 'Fall', 'Winter']
season = pd.Series(data)
print(season, '\n')

print(season.iloc[2])
```

```python
import pandas as pd

score = pd.DataFrame([[85, 96, 40, 95], [73, 69, 45, 80], [78, 50, 60, 90]])

print(score, '\n')
print(type(score), '\n')

print(score.index, '\n')
print(score.columns, '\n')

print(score.iloc[1, 2])
```

```python
import pandas as pd

age = pd.Series([25, 34, 19, 45])
print(age, '\n')

age.index = ['John', 'Jane', 'Tom', 'Luka']
print(age, '\n')

print(age.iloc[2], '\n')
print(age.loc['Tom'], '\n')

score = pd.DataFrame([[85, 96, 40, 95], [73, 69, 45, 80], [78, 50, 60, 90]])
print(score, '\n')

score.index = ['John', 'Jane', 'Tom']
score.columns = ['Kor', 'Eng', 'Math', 'Sci']
print(score, '\n')

print(score.iloc[2, 1], '\n')
print(score.loc['Tom', 'Eng'])
```
