```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = sns.load_dataset('penguins')
df = df.dropna()
print(df)

#1
df.plot.scatter(x = 'bill_length_mm', y = 'bill_depth_mm')
plt.show()

#2
print(df[['bill_length_mm', 'bill_depth_mm']].corr(), '\n')

#3
df.plot.scatter(x = "bill_length_mm", y = "flipper_length_mm")
plt.show()

#4
print(df[["bill_length_mm", "flipper_length_mm"]].corr(), '\n')

#5
print(df.loc[:, 'species'].unique(), '\n')

#6
dict = {'Adelie' : 'red', 'Chinstrap' : 'green', 'Gentoo' : 'blue'}
color = list(dict[key] for key in df.species)
df.plot.scatter(x = "bill_length_mm", y = "bill_depth_mm", c = color)
plt.show()

#8
dict2 = {'Male' : 'blue', 'Female' : 'red'}
color2 = list(dict2[key] for key in df.sex)
df.plot.scatter(x = 'bill_length_mm', y = 'bill_depth_mm', c = color2)
plt.show()

#10
data = df.loc[:, ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']]
pd.plotting.scatter_matrix(data)
plt.show()

#11
print(data.corr(), '\n')

#12
```
