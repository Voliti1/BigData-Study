```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = sns.load_dataset('penguins')
df = df.dropna()
print(df)

#1
df.plot.scatter(x = 'bill_length_mm', y = 'bill_depth_mm')
plt.show()

#2
print(df[['bill_length_mm', 'bill_depth_mm']].corr(), '\n')

#3
df.plot.scatter(x = "bill_length_mm", y = "flipper_length_mm")
plt.show()

#4
print(df[["bill_length_mm", "flipper_length_mm"]].corr())

#5
print(df.loc[:, 'species'].unique())
```
