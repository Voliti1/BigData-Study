```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

#1
df = sns.load_dataset('tips')
day = df.day

#2
print(day, '\n')
print(day.dtype, '\n')

#3
fd = day.value_counts()

#4
print(fd / len(day))

#5
fd.plot.bar(xlabel = "day", ylabel = "frequency", title = "frequency of day", rot = 360)
plt.show()

#6
fd.plot.barh(xlabel = "frequency", ylabel = 'day', color = "r")
plt.show()

#7
fd.plot.pie(xlabel = "day", autopct = '%1.0f%%')
plt.show()

#8
smoker = df.smoker.value_counts()
smoker.plot.bar(xlabel = "smoker", ylabel = "num", title = "num of smoker")
plt.show()

#9
time = df.time.value_counts()
time.plot.pie(xlabel = "num", autopct = "%1.0f%%")

#10
df = sns.load_dataset('tips')
amount = df.total_bill

#11
print(amount, '\n')
print(amount.dtype, '\n')

#12
print(amount.mean(), '\n')
print(amount.median(), '\n')
print(amount.std(), '\n')

#13
print(amount.quantile([0.25, 0.5, 0.75]), '\n')

#14
amount.plot.hist(bins = 6)
plt.show()

#15
print(amount.value_counts(bins = 6).sort_index())

#16
amount.plot.box()

#17
df.boxplot(column = 'total_bill', by = 'time')

#18
fig, axes  = plt.subplots(2, 2)
amount.plot.hist(bins = 6, ax = axes[0, 0])
amount.plot.box(ax = axes[0, 1])
fd.plot.bar(xlabel = "day", ylabel = "frequency", title = "frequency of day", rot = 360, ax = axes[1, 0])
fd.plot.pie(xlabel = "day", autopct = '%1.0f%%', ax = axes[1, 1])
plt.suptitle("그래프 총 집합")
plt.show()
```
