```python
import pandas as pd
import numpy as np

temp = pd.Series([-0.8, -0.1, 7.7, 13.8, 18.0, 22.4,
                    25.9, 25.3, 21.0, 14.0, 9.6, -1.4])

print(temp, '\n')
print(temp.index, '\n')

temp.index = ['1월', '2월', '3월', '4월', '5월', '6월',
              '7월', '8월', '9월', '10월', '11월', '12월']

print(temp)

print(temp.size, '\n')
print(len(temp), '\n')
print(temp.shape, '\n')
print(temp.dtype, '\n')

print(temp.iloc[2], '\n')
print(temp.loc['3월'], '\n')
print(temp.iloc[[3, 5, 7]], '\n')
print(temp.loc[['4월', '6월', '8월']], '\n')
print(temp.iloc[5:8], '\n')
print(temp.loc['6월':'9월'], '\n')
print(temp.iloc[:4], '\n')
print(temp.iloc[9:], '\n')
print(temp.iloc[:], '\n')

print(temp.loc[temp >= 15], '\n')
print(temp.loc[(temp >= 15) & (temp < 25)], '\n')
print(temp.loc[(temp < 5) | (temp >= 25)], '\n')

march = temp.loc['3월']
print(temp.loc[temp < march], '\n')

print(temp.where(temp >= 15), '\n')
print(temp.where(temp >= 15).dropna(), '\n')


print(temp.loc[temp >= 15], '\n')
print(temp.loc[temp >= 15].index, '\n')

print(temp + 1, '\n')
print(2 * temp + 0.1, '\n')
print(temp + temp, '\n')
print(temp.loc[temp >= 15] + 1, '\n')

print(temp.sum())
print(temp.mean())
print(temp.median())
print(temp.max())
print(temp.min())
print(temp.std())
print(temp.var())
print(temp.abs())
print(temp.describe())
```

```python
import pandas as pd

salary = pd.Series([20, 15, 18, 30])
score = pd.Series([75, 80, 90, 60], index = ['Kor', 'Eng', 'Math', 'Soc'])

print(salary, '\n')
print(score, '\n')

score.iloc[0] = 85
print(score, '\n')

score.loc['Soc'] = 65
print(score, '\n')

score.loc[['Eng', 'Math']] = [70, 80]
print(score, '\n')

score.loc['Phy'] = 50
print(score, '\n')
# score.iloc[5] = 90

next_idx = salary.size
#salary.iloc[next_idx] = 33
salary.loc[next_idx] = 33
print(salary, '\n')

score = pd.Series([85, 70, 80, 65, 50, 95], index = ['Kor', 'Eng', 'Math', 'Soc', 'Phy', 'Mus'])
score.drop('Phy')
print(score, '\n')
score = score.drop('Phy')
print(score)
```
